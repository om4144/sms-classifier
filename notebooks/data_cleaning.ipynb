### Step 4: Categorical Encoding

---

#### 💡 What is Encoding?
Encoding is the process of converting categorical text data into numerical values so that machine learning algorithms can process and learn from the dataset.

---

#### 📌 Common Encoding Techniques

##### 1. One-Hot Encoding
Creates a new binary column (`0` or `1`) for each unique category.
* **Example:**
  | Input Category | Red | Blue | Green |
  | :--- | :---: | :---: | :---: |
  | **Red** | 1 | 0 | 0 |
  | **Blue** | 0 | 1 | 0 |
  | **Green** | 0 | 0 | 1 |

##### 2. Ordinal Encoding
Assigns ordered numeric values (e.g., 1, 2, 3) based on a logical rank or hierarchy.
* **Example:** `Difficulty = ['low', 'medium', 'high']`
  * `High` → 1
  * `Medium` → 2
  * `Low` → 3

##### 3. Label Encoding
Assigns a unique integer identifier to each unique category.
> **Note:** Primarily applied to **target columns** in classification problems.
* **Example:** `Difficulty = ['low', 'medium', 'high']`
  * `High` → 0
  * `Medium` → 1
  * `Low` → 2

---

#### 🎯 Strategy for Our Model
We will use **Label Encoding** to transform our target column:
* `ham` → `0`
* `spam` → `1`

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df["target"] = le.fit_transform(df["target"])
```
